```python
import pandas as pd

df = pd.read_csv('мои сделки - Лист1.csv')
df.columns = df.columns.str.strip().str.lower()

for col in ['количество', 'цена', 'комиссия']:
    df[col] = pd.to_numeric(df[col].astype(str).str.replace(',', '.').str.replace(r'\s+', '', regex=True), errors='coerce').fillna(0)

df['сумма'] = df['количество'] * df['цена'] + df['комиссия']
result = df.groupby(['тикер', 'валюта'])['сумма'].sum()

print("Сумма вложений по каждому активу:")
print(result)
```

```python
def pnl(buy, now, qty):
    profit = (now - buy) * qty
    percent = ((now - buy) / buy) * 100
    return round(profit, 2), round(percent, 2)

positions = [
    {"ticker": "SPUS", "buy": 57.63, "now": 55.78, "qty": 27, "cur": "USD"},
    {"ticker": "VXUS", "buy": 64.35, "now": 62.50, "qty": 9, "cur": "USD"},
    {"ticker": "SCHD", "buy": 27.03, "now": 28.40, "qty": 20, "cur": "USD"},
    {"ticker": "HSBK", "buy": 371.81, "now": 372.48, "qty": 101, "cur": "KZT"},
    {"ticker": "KZTO", "buy": 1373.0, "now": 1379.89, "qty": 9, "cur": "KZT"}
]

for item in positions:
    money, pct = pnl(item["buy"], item["now"], item["qty"])
    print(f"{item['ticker']}: {money} {item['cur']} ({pct}%)")
```

```python
import pandas as pd
import yfinance as yf

# 1. Создаём таблицу с твоими сделками из Google Sheets
portfolio = pd.DataFrame({
    "Тикер": ["SPUS", "VXUS", "SCHD"],
    "Количество":[27, 7, 15],
    "Цена покупки ($)": [57.63, 82.73, 36.04]
})

# 2. Автоматически подтягиваем текущие цены с биржи
current_prices = []
for ticker in portfolio["Тикер"]:
    data = yf.download(ticker, period="5d", auto_adjust=True, progress=False)
    price = float(data["Close"].iloc[-1].iloc[0])
    current_prices.append(round(price, 2))

portfolio["Текущая цена ($)"] = current_prices

# 3. Добавляем два новых столбца с расчётом прибыли
portfolio["Прибыль ($)"] = ((portfolio["Текущая цена ($)"] - portfolio["Цена покупки ($)"]) * portfolio["Количество"]).round(2)
portfolio["Прибыль (%)"] = (((portfolio["Текущая цена ($)"] - portfolio["Цена покупки ($)"]) / portfolio["Цена покупки ($)"]) * 100).round(2)

# Красиво выводим таблицу
display(portfolio)
```

```python
import yfinance as yf
import matplotlib.pyplot as plt

# Скачиваем данные за 3 месяца для акций и крипты
data = yf.download(["SPUS", "SCHD", "BTC-USD", "SOL-USD"], period="3mo", auto_adjust=True, progress=False)["Close"]

# Заполняем пропуски (выходные дни, когда биржа акций закрыта, а крипта торгуется)
data = data.ffill().dropna()

# Собираем два портфеля (50/50 внутри каждого) и нормируем их к 50 на старте
stocks_portfolio = ((data["SPUS"] / data["SPUS"].iloc[0]) * 0.5 + (data["SCHD"] / data["SCHD"].iloc[0]) * 0.5) * 50
crypto_portfolio = ((data["BTC-USD"] / data["BTC-USD"].iloc[0]) * 0.5 + (data["SOL-USD"] / data["SOL-USD"].iloc[0]) * 0.5) * 50

# Рисуем линейный график
plt.figure(figsize=(10, 5))
plt.plot(stocks_portfolio.index, stocks_portfolio, label="Портфель акций (SPUS + SCHD)", color="blue", linewidth=2)
plt.plot(crypto_portfolio.index, crypto_portfolio, label="Криптопортфель (BTC + SOL)", color="orange", linewidth=2)

plt.axhline(50, color="gray", linestyle="--", alpha=0.7)
plt.title("Сравнение доходности двух портфелей (старт = 50)")
plt.ylabel("Условная стоимость ($)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import yfinance as yf
import pandas as pd

# 1. Твои позиции из таблицы
portfolio = pd.DataFrame({
    "Тикер": ["SPUS", "VXUS", "SCHD"],
    "Цена покупки": [57.63, 82.73, 36.04]
})

# 2. Скачиваем цены за последний месяц (30 дней)
prices = yf.download(list(portfolio["Тикер"]), period="1mo", auto_adjust=True, progress=False)["Close"]

# 3. Считаем текущую доходность (%) и волатильность (%) для каждого актива
returns_pct = {}
volatility_pct = {}

for idx, row in portfolio.iterrows():
    ticker = row["Тикер"]
    buy_price = row["Цена покупки"]
    now_price = float(prices[ticker].iloc[-1])

    # Прибыль в % от цены покупки
    returns_pct[ticker] = ((now_price - buy_price) / buy_price) * 100
    # Дневная волатильность за месяц в %
    volatility_pct[ticker] = float(prices[ticker].pct_change().dropna().std()) * 100

# 4. Находим лидера по доходности и самый волатильный актив
best_ticker = max(returns_pct, key=returns_pct.get)
most_volatile = max(volatility_pct, key=volatility_pct.get)

# 5. Автоматический вывод одной строкой
print(f"Итог по портфелю: самая прибыльная позиция — {best_ticker} ({round(returns_pct[best_ticker], 2)}%), самая волатильная — {most_volatile} ({round(volatility_pct[most_volatile], 2)}% в день).")
```
